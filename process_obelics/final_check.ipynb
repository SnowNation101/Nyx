```python
import json
with open("mm_dataset.json", "r") as f:
    data = json.load(f)
```

```python
print(json.dumps(data[0], indent=2))
```

```python
for item in data:
    if item['qry'].count('<|image_1|>') != len(item['qry_image_path']):
        print(f"Mismatch in qry_images for item: {item['qry']}")
    if item['pos_text'].count('<|image_1|>') != len(item['pos_image_path']):
        print(f"Mismatch in pos_images for item: {item['pos_text']}")
    for neg_text, neg_image_path in zip(item['neg_text'], item['neg_image_path']):
        if neg_text.count('<|image_1|>') != len(neg_image_path):
            print(f"Mismatch in neg_images for item: {neg_text}")
```
